````python
import os
import asyncio
import nest_asyncio
!pip install playwright
!pip install PyPDF2
from playwright.async_api import async_playwright
from PyPDF2 import PdfMerger

# This allows Playwright's async functions to run smoothly inside Colab's notebook environment
nest_asyncio.apply()

# Configuration
URL = "https://dummytestsite.lasu.edu.ng/fake_studentid/index.php"
START_MATRIC = 240591001
END_MATRIC = 240591445

# Temporary Colab storage for individual PDFs (fast, gets deleted when Colab closes)
OUTPUT_DIR = "/content/downloaded_ids"

# Save the final merged file directly to your Google Drive
# You can change 'MyDrive' to a specific folder path like 'MyDrive/School_Project/Merged_IDs.pdf'
FINAL_PDF = "/content/drive/MyDrive/Merged_Course_Mates_IDs.pdf"

async def generate_and_download_ids():
    if not os.path.exists(OUTPUT_DIR):
        os.makedirs(OUTPUT_DIR)

    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=True)
        page = await browser.new_page()

        for matric in range(START_MATRIC, END_MATRIC + 1):
            matric_str = str(matric)
            print(f"Processing Matric Number: {matric_str}...")

            try:
                await page.goto(URL)
                await page.fill("input[type='text'], input[name='matric_no']", matric_str)
                await page.click("button:has-text('GENERATE'), input[value='GENERATE'], input[type='submit']")
                await page.wait_for_load_state('networkidle')

                content = await page.content()
                if "Invalid Matric Number" in content:
                    print(f" --> Skipping {matric_str}: Invalid Matric Number found.")
                    continue

                pdf_path = os.path.join(OUTPUT_DIR, f"{matric_str}.pdf")

                await page.evaluate("""
                    var buttons = document.querySelectorAll('button, a, input[type="button"]');
                    buttons.forEach(btn => {
                        if(btn.innerText.includes('Print') || btn.innerText.includes('back')) {
                            btn.style.display = 'none';
                        }
                    });
                """)

                await page.pdf(path=pdf_path, format="A4", print_background=True)
                print(f" --> Successfully saved: {pdf_path}")

            except Exception as e:
                print(f" --> Error processing {matric_str}: {e}")

        await browser.close()

def merge_pdfs():
    print("\nStarting PDF merge process...")
    if not os.path.exists(OUTPUT_DIR):
        print("No downloaded PDFs found to merge.")
        return

    merger = PdfMerger()
    pdf_files = sorted([f for f in os.listdir(OUTPUT_DIR) if f.endswith('.pdf')])

    if not pdf_files:
        print("No valid PDFs were downloaded.")
        return

    for pdf in pdf_files:
        file_path = os.path.join(OUTPUT_DIR, pdf)
        # Grab only the first page (index 0)
        merger.append(file_path, pages=(0, 1))

    merger.write(FINAL_PDF)
    merger.close()
    print(f"Success! The first pages of all IDs have been saved to your Google Drive at: {FINAL_PDF}")

async def main():
    print("--- Starting ID Card Generation ---")
    await generate_and_download_ids()
    print("--- Download Phase Complete ---")
    merge_pdfs()

# In Jupyter/Colab, we use asyncio.get_event_loop().run_until_complete()
# or simply await the main function directly.
await main()
````

### The Rescue Script

Create a **new code cell** at the bottom of your notebook, paste this code, and run it. It skips the download phase completely and just merges the files we already have.
